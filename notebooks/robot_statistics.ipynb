```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import pearsonr
```

```python
df = pd.read_csv('../data/robot_runs.csv')

print("Перші записи DataFrame:")
display(df.head())

print("-" * 50)
print(f"Кількість рядків: {len(df)}")
print("-" * 50)
print("Назви стовпців:")
print(df.columns.tolist())
print("-" * 50)
print("Типи даних:")
print(df.dtypes)
```

```python
mean_battery = df["battery_before_percent"].mean()
print(f"Середній заряд акумулятора перед запуском: {mean_battery:.2f}%")
```

```python
mean_route_time = df["route_time_sec"].mean()
print(f"Середній час проходження маршруту: {mean_route_time:.2f} сек")
```

```python
mean_package_weight = df["package_weight_kg"].mean()
print(f"Середня вага вантажу: {mean_package_weight:.2f} кг")
```

```python
median_battery = df["battery_before_percent"].median()
print(f"Медіана заряду акумулятора: {median_battery}%")
```

```python
median_route_time = df["route_time_sec"].median()
print(f"Медіана часу проходження маршруту: {median_route_time} сек")
```

```python
median_package_weight = df["package_weight_kg"].median()
print(f"Медіана ваги вантажу: {median_package_weight} кг")
```

```python
mode_battery = df["battery_before_percent"].mode()
print("Мода (найчастіші значення) заряду акумулятора:")
print(mode_battery.tolist())
```

```python
mode_route_time = df["route_time_sec"].mode()
print("Мода часу проходження маршруту:")
print(mode_route_time.tolist())
```

```python
mode_package_weight = df["package_weight_kg"].mode()
print("Мода ваги вантажу:")
print(mode_package_weight.tolist())
```

```python
variance_battery = df["battery_before_percent"].var(ddof=0)
print(f"Дисперсія генеральної сукупності для заряду акумулятора (ddof=0): {variance_battery:.2f}")
```

```python
variance_route_time = df["route_time_sec"].var(ddof=0)
print(f"Дисперсія генеральної сукупності для часу маршруту (ddof=0): {variance_route_time:.2f}")
```

```python
variance_package_weight = df["package_weight_kg"].var(ddof=0)
print(f"Дисперсія генеральної сукупності для ваги вантажу (ddof=0): {variance_package_weight:.4f}")
```

```python
std_battery = df["battery_before_percent"].std(ddof=0)
print(f"Стандартне відхилення генеральної сукупності для заряду батареї: {std_battery:.2f}%")
```

```python
std_route_time = df["route_time_sec"].std(ddof=0)
print(f"Стандартне відхилення генеральної сукупності для часу маршруту: {std_route_time:.2f} сек")
```

```python
std_package_weight = df["package_weight_kg"].std(ddof=0)
print(f"Стандартне відхилення генеральної сукупності для ваги вантажу: {std_package_weight:.2f} кг")
```

```python
cov_weight_time = df["package_weight_kg"].cov(df["route_time_sec"])
print(f"Коваріація між вагою вантажу та часом маршруту: {cov_weight_time:.2f}")
```

```python
corr_result = stats.pearsonr(df["package_weight_kg"], df["route_time_sec"])

print(f"Коефіцієнт кореляції Пірсона (r): {corr_result.statistic:.4f}")
print(f"Статистична значущість (p-value): {corr_result.pvalue}")
```

```python
sample = df.sample(n=10, random_state=42)
print("Отримана вибірка з 10 запусків:")
display(sample)
```

```python
sample_mean_battery = sample["battery_before_percent"].mean()
sample_mean_route_time = sample["route_time_sec"].mean()
sample_mean_package_weight = sample["package_weight_kg"].mean()

print(f"Вибіркове середнє заряду акумулятора: {sample_mean_battery:.2f}%")
print(f"Вибіркове середнє часу маршруту: {sample_mean_route_time:.2f} сек")
print(f"Вибіркове середнє ваги вантажу: {sample_mean_package_weight:.2f} кг")
```

```python
sample_variance_battery = sample["battery_before_percent"].var(ddof=1)
sample_variance_route_time = sample["route_time_sec"].var(ddof=1)
sample_variance_package_weight = sample["package_weight_kg"].var(ddof=1)

print(f"Вибіркова дисперсія акумулятора: {sample_variance_battery:.2f}")
print(f"Вибіркова дисперсія часу маршруту: {sample_variance_route_time:.2f}")
print(f"Вибіркова дисперсія ваги вантажу: {sample_variance_package_weight:.4f}")
```

```python
sample_std_battery = sample["battery_before_percent"].std(ddof=1)
sample_std_route_time = sample["route_time_sec"].std(ddof=1)
sample_std_package_weight = sample["package_weight_kg"].std(ddof=1)

print(f"Вибіркове ст. відхилення акумулятора: {sample_std_battery:.2f}%")
print(f"Вибіркове ст. відхилення часу маршруту: {sample_std_route_time:.2f} сек")
print(f"Вибіркове ст. відхилення ваги вантажу: {sample_std_package_weight:.2f} кг")
```

```python
summary_data = {
    "Показник": [
        "Середній заряд акумулятора (%)", "Дисперсія заряду", "Стандартне відхилення заряду",
        "Середній час маршруту (сек)", "Дисперсія часу", "Стандартне відхилення часу",
        "Середня вага вантажу (кг)", "Дисперсія ваги", "Стандартне відхилення ваги"
    ],
    "Генеральна сукупність": [
        mean_battery, variance_battery, std_battery,
        mean_route_time, variance_route_time, std_route_time,
        mean_package_weight, variance_package_weight, std_package_weight
    ],
    "Вибірка (n=10)": [
        sample_mean_battery, sample_variance_battery, sample_std_battery,
        sample_mean_route_time, sample_variance_route_time, sample_std_route_time,
        sample_mean_package_weight, sample_variance_package_weight, sample_std_package_weight
    ]
}

df_comparison = pd.DataFrame(summary_data).round(4)
print("Порівняння статистичних показників:")
display(df_comparison)
```
